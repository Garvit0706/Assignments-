Q1. What is Gradient Boosting Regression?
Gradient Boosting Regression is an ensemble technique in machine learning that builds a model by sequentially adding weak learners (usually decision trees) to minimize a loss function. In the context of regression, the goal is to predict a continuous target variable by gradually improving the model's predictions through the use of gradient descent. Each weak learner is trained to correct the residual errors of the combined ensemble of previous learners.

```python
2.
import numpy as np
from sklearn.metrics import mean_squared_error, r2_score

class GradientBoostingRegressor:
    def __init__(self, n_estimators=100, learning_rate=0.1, max_depth=3):
        self.n_estimators = n_estimators
        self.learning_rate = learning_rate
        self.max_depth = max_depth
        self.models = []
        self.initial_pred = None
    
    def fit(self, X, y):
        # Initialize the predictions with the mean of y
        self.initial_pred = np.mean(y)
        y_pred = np.full(y.shape, self.initial_pred)
        
        for _ in range(self.n_estimators):
            # Compute the residuals
            residuals = y - y_pred
            
            # Fit a simple decision tree to the residuals
            tree = self._fit_tree(X, residuals)
            self.models.append(tree)
            
            # Update the predictions
            y_pred += self.learning_rate * tree.predict(X)
    
    def predict(self, X):
        y_pred = np.full((X.shape[0],), self.initial_pred)
        for tree in self.models:
            y_pred += self.learning_rate * tree.predict(X)
        return y_pred
    
    def _fit_tree(self, X, y):
        from sklearn.tree import DecisionTreeRegressor
        tree = DecisionTreeRegressor(max_depth=self.max_depth)
        tree.fit(X, y)
        return tree

# Generate a small dataset
np.random.seed(42)
X = np.random.rand(100, 1) * 10
y = 2 * X.squeeze() + 1 + np.random.randn(100) * 0.5

# Train the gradient boosting model
gbr = GradientBoostingRegressor(n_estimators=50, learning_rate=0.1, max_depth=3)
gbr.fit(X, y)

# Predict on the training data
y_pred = gbr.predict(X)

# Evaluate the model
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f'Mean Squared Error: {mse:.4f}')
print(f'R-squared: {r2:.4f}')
```

```python
3.
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [50, 100, 200],
    'learning_rate': [0.01, 0.1, 0.2],
    'max_depth': [2, 3, 4]
}

# Create a basic Gradient Boosting model
gbr = GradientBoostingRegressor()

# Perform grid search
grid_search = GridSearchCV(estimator=gbr, param_grid=param_grid, cv=5, scoring='r2')
grid_search.fit(X, y)

# Output the best parameters
print(f'Best Parameters: {grid_search.best_params_}')

# Evaluate the best model
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X)

mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f'Best Model Mean Squared Error: {mse:.4f}')
print(f'Best Model R-squared: {r2:.4f}')
```

Q4. What Is a Weak Learner in Gradient Boosting?
A weak learner in Gradient Boosting is a simple model that performs slightly better than random guessing. Typically, a weak learner is a shallow decision tree (often referred to as a decision stump) that is only a few levels deep. The idea is to use many weak learners, each correcting the errors of the previous ones, to build a strong predictive model.

Q5. What Is the Intuition Behind the Gradient Boosting Algorithm?
The intuition behind Gradient Boosting is to build a powerful predictive model by sequentially adding models that correct the errors of the combined ensemble of all previous models. Each model is trained on the residuals (errors) of the current prediction, with the goal of reducing these errors in the next iteration. The process is similar to gradient descent, where the algorithm moves in the direction of the negative gradient of the loss function to minimize it.


Q6. How Does the Gradient Boosting Algorithm Build an Ensemble of Weak Learners?
The Gradient Boosting algorithm builds an ensemble of weak learners by following these steps:
Initialize: Start with an initial prediction (often the mean of the target variable).
Calculate Residuals: Compute the residuals (differences between the true values and the current predictions).
Train Weak Learner: Train a weak learner (e.g., a decision tree) on the residuals.
Update Predictions: Update the predictions by adding the output of the weak learner, scaled by a learning rate.
Iterate: Repeat the process for a predefined number of iterations or until the error is minimized.


Q7. What Are the Steps Involved in Constructing the Mathematical Intuition of the Gradient Boosting Algorithm?
The steps involved in constructing the mathematical intuition of the Gradient Boosting algorithm are:
Define the Loss Function: Choose a differentiable loss function that quantifies the error between the predicted and actual values (e.g., Mean Squared Error for regression).
Initialize the Model: Start with a simple model that predicts the mean of the target variable.
Compute the Residuals: Calculate the negative gradient of the loss function with respect to the predictions. This represents the residuals that the next model needs to correct.
Fit a Weak Learner: Train a weak learner on the residuals to predict the direction and magnitude of the error.
Update the Model: Add the output of the weak learner to the current predictions, scaled by the learning rate.
Iterate: Continue adding weak learners, each focused on reducing the residuals, until the model converges to a minimum loss or a maximum number of iterations is reached.
