Q3. How Does Bernoulli Naive Bayes Handle Missing Values?
Bernoulli Naive Bayes, like other variants of Naive Bayes, does not inherently handle missing values. If a feature value is missing, the algorithm might assume that the feature is absent (0) if not handled explicitly. However, it is common practice to preprocess the data by filling in missing values (e.g., with mean, median, mode, or a constant value) before applying the Bernoulli Naive Bayes classifier.

Q4. Can Gaussian Naive Bayes be Used for Multi-Class Classification?
Yes, Gaussian Naive Bayes can be used for multi-class classification. In the multi-class setting, Gaussian Naive Bayes assumes that each class has its own normal distribution for each feature, and it computes the probability of a data point belonging to each class. The class with the highest posterior probability is then chosen as the predicted class.

```python
5.
import pandas as pd
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.naive_bayes import BernoulliNB, MultinomialNB, GaussianNB
from sklearn.preprocessing import StandardScaler

# Load the dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/spambase/spambase.data"
data = pd.read_csv(url, header=None)

# Split features and target
X = data.iloc[:, :-1]  # Features
y = data.iloc[:, -1]   # Target

# Optional: Normalize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Define classifiers
classifiers = {
    'BernoulliNB': BernoulliNB(),
    'MultinomialNB': MultinomialNB(),
    'GaussianNB': GaussianNB()
}

# Evaluate each classifier using cross-validation
for name, clf in classifiers.items():
    print(f"Evaluating {name}...")
    accuracy = cross_val_score(clf, X_scaled, y, cv=10, scoring='accuracy').mean()
    precision = cross_val_score(clf, X_scaled, y, cv=10, scoring='precision').mean()
    recall = cross_val_score(clf, X_scaled, y, cv=10, scoring='recall').mean()
    f1 = cross_val_score(clf, X_scaled, y, cv=10, scoring='f1').mean()
    print(f"Accuracy: {accuracy:.4f}, Precision: {precision:.4f}, Recall: {recall:.4f}, F1 Score: {f1:.4f}\n")
```
