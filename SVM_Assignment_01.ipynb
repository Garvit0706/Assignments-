```python
Q4. What is the Role of Support Vectors in SVM? Explain with Example
Support vectors are the data points that are closest to the hyperplane and are critical in defining the position and 
orientation of the hyperplane. They are the "supporting" elements that help the SVM determine the decision boundary.

Example:

Consider a simple 2D classification problem where two classes are linearly separable. The SVM algorithm will find the optimal 
hyperplane that separates the two classes. The data points that lie closest to the hyperplane (on the boundary of the margin) 
are the support vectors.

These support vectors are the only points that influence the position of the hyperplane. Even if other data points are moved, 
the hyperplane remains unchanged as long as the support vectors do not move.

Role:

Support vectors are the key data points that determine the decision boundary.
They define the margin around the hyperplane, ensuring the classifier is robust.
In many cases, only a small fraction of the total data points become support vectors, making SVMs efficient in handling large datasets.
```

Q5. Illustrate with Examples and Graphs of Hyperplane, Marginal Plane, Soft Margin, and Hard Margin in SVM
Hyperplane and Marginal Plane
Hyperplane: The line or plane that separates the classes.
Marginal Plane: The planes parallel to the hyperplane that pass through the support vectors and define the margin.
Graphical Example:

In a 2D space, the hyperplane is a line, and the margin is the distance between the two marginal planes.
Hard Margin SVM
Hard Margin: Assumes that the data is perfectly linearly separable. No data points are allowed to lie within the margin or on the wrong side of the hyperplane.
Example: This is suitable for situations where there are no outliers or noise in the data.
Graphical Example:

All data points lie on the correct side of the margin, with no violations.
Soft Margin SVM
Soft Margin: Allows some data points to lie within the margin or on the wrong side of the hyperplane to account for outliers or noise. This is controlled by the regularization parameter 
𝐶
C.
Example: This is suitable for real-world data where perfect separation is not possible.
Graphical Example:

Some data points lie within the margin or on the wrong side of the hyperplane, but the overall margin is maximized.

```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import numpy as np

# Load the iris dataset
iris = datasets.load_iris()
X = iris.data[:, :2]  # Taking only two features for simplicity
y = iris.target

# Only consider two classes for binary classification
X = X[y != 2]
y = y[y != 2]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = SVC(kernel='linear', C=1.0)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%")


# Create a mesh to plot in
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                     np.arange(y_min, y_max, 0.02))

Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.8)
plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', marker='o')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.title('SVM Decision Boundary')
plt.show()


C_values = [0.1, 1, 10]
for C in C_values:
    model = SVC(kernel='linear', C=C)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    print(f"C={C}: Accuracy: {accuracy * 100:.2f}%")
```
